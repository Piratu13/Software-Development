```python

# STUDENT GRADING SYSTEM
# PART 3C - DEFINE THE CATEGORICAL ROUNDING FUNCTION

# Define a function that receives the student's overall score.
def round_to_category(score):

    # Store all permitted university marks.
    allowed_marks = [
        0, 5, 15, 25, 32, 35, 38,
        42, 45, 48, 52, 55, 58,
        62, 65, 68, 72, 75, 78,
        82, 85, 92, 100
    ]

    # Find the permitted mark closest to the student's score.
    # abs() measures the distance between two numbers.
    # min() selects the candidate with the smallest distance.
    # -candidate_mark selects the higher mark when distances are equal.
    nearest_mark = min(allowed_marks, key=lambda candidate_mark: (abs(score - candidate_mark), -candidate_mark))

# TOUPLE RETURN

    # Find the grading category corresponding to the rounded mark.
    category = determine_category(nearest_mark)

    # Return both the rounded mark and category as a tuple.
    return nearest_mark, category
```

```python

# PART 3B - DETERMINE THE STUDENT'S GRADING CATEGORY

# Define a function that determines the category from a rounded mark.
def determine_category(rounded_mark):
    
# The rounded_mark variable was calculated in Cell 2.
# Check which category contains that permitted mark.

# Check for Gold Standard.
    if rounded_mark == 100:
        return "Gold Standard"

# Check for Upper First.
    elif rounded_mark == 82 or rounded_mark == 85 or rounded_mark == 92:
        return "Upper First"

# Check for First.
    elif rounded_mark == 72 or rounded_mark == 75 or rounded_mark == 78:
        return "First"

# Check for Upper Second.
    elif rounded_mark == 62 or rounded_mark == 65 or rounded_mark == 68:
        return "Upper Second"

# Check for Lower Second.
    elif rounded_mark == 52 or rounded_mark == 55 or rounded_mark == 58:
        return "Lower Second"

# Check for Third.
    elif rounded_mark == 42 or rounded_mark == 45 or rounded_mark == 48:
        return "Third"

# Check for Condonable Fail.
    elif rounded_mark == 32 or rounded_mark == 35 or rounded_mark == 38:
        return "Condonable Fail"

# Check for Fail.
    elif rounded_mark == 5 or rounded_mark == 15 or rounded_mark == 25:
        return "Fail"

# Check for No Submission.
    elif rounded_mark == 0:
        return "No Submission"

# If no permitted mark matches, display an error message.
    else:
        print("Invalid Mark")
        return "Invalid Mark"
```

```python

# PART 3E - ASSESSMENT SCORE VALIDATION

# Define a reusable function for entering assessment scores.
def get_valid_score(message):

    # Keep asking untill a valid score is returned.
    while True:
        try:
            # Ask for a score and convert it into a decimal number.
            score = float(input(message))

            # Accept and return the score if it is between 0 and 100.
            if 0 <= score <= 100:
                # Return the accepted score.
                return score
            # Reject values outside the permitted range.
            print("The input you entered was invalid")
            print("Please enter a score between 0 and 100.")
        except ValueError:
            # Handle text that cannot be converted into a number.
            print("The input you entered was invalid")
            print("Please enter a numeric score.")

          
  
```

```python
 
# PART 3F - CALCULATE THE OVERALL SCORE

# Define a function that receives four assessment scores.
def calculate_overall_score(coursework1, coursework2, coursework3, final_exam):

    # Multiply each score by its assessment weight.
    overall_score = (
        coursework1 * 0.10 + coursework2 * 0.40
        + coursework3 * 0.25 + final_exam * 0.25
        )

    # Return the calculated score to the calling code.
    return overall_score
```

```python

# PART 3J - STUDENT FULL NAME VALIDATION

# Define a reusable function to collect and validate a full name.
def get_student_name():

    # Keep requesting a name until at least two words are entered.
    while True:

        # Ask for the student's full name.
        # strip() removes spaces at the beginning and end.
        student_name = input("Enter student's full name: ").strip()

        # split() separates the name into individual words.
        name_parts = student_name.split()

        # Check whether the name contains at least two words.
        if len(name_parts) < 2:
            print("The input you entered was invalid")
            print("Please enter a first name and surname.")
            continue

        # Check whether every name part contains only letters.
        if not all(part.isalpha() for part in name_parts):
            print("The input you entered was invalid")
            print("Names must contain letters only.")
            continue

        # Capitalise the first letter of each word.
        student_name = " ".join(name_parts).title()

        # Return the validated full name to main().
        return student_name

       
    
```

```python

# PART 3K - STUDENT DATE OF BIRTH VALIDATION

# FUNCTION 1: Validate the student's date of birth.

# Define a function to collect and validate a student's date of birth.
def get_student_dob():

    # Keep requesting a date until a valid date is entered.
    while True:

        # Ask the professor to enter the student's date of birth.
        date_text = input("Enter date of birth (YYYY-MM-DD): ").strip()

        try:
            # Convert the entered text into a Python date object.
            birth_date = datetime.strptime(date_text, "%Y-%m-%d").date()

            # Check that the entered date uses the exact ISO format.
            if birth_date.strftime("%Y-%m-%d") != date_text:
                print("The input you entered was invalid")
                continue

            # Reject dates that are later than today's date.
            if birth_date > date.today():
                print("The input you entered was invalid")
                continue

            # Calculate the student's age using our existing function.
            student_age = calculate_student_age(birth_date)

            # Reject students who are younger than 16.
            if student_age < 16:
                print("The input you entered was invalid")
                print("Student must be at least 16 years old.")
                continue


            # Return the validated date to the calling function.
            return birth_date

        # Handle impossible dates and incorrect date input.
        except ValueError:

            # Display the warning required by the coursework.
            print("The input you entered was invalid")


# FUNCTION 2: Calculate the student's age.
def calculate_student_age(birth_date):

    # Retrieve today's date.
    today = date.today()

    # Calculate the difference between the years.
    age = today.year - birth_date.year

    # Check whether the birthday is still upcoming.
    birthday_not_passed = (
        (today.month, today.day) <
        (birth_date.month, birth_date.day)
    )

    # Subtract one if the birthday hasn't happened.
    if birthday_not_passed:
        age = age - 1

    # Return the calculated age.
    return age
```

```python
# PART 3G / 3H / 3I - MAIN STUDENT GRADING PROGRAM

def main():

    # Create an empty list to store all student records.
    students = []

    # Process a maximum of three students.
    for student_number in range(3):

        # Ask for a student ID or the command 'end'.
        student_id = input("Enter student ID (2 digits), or 'end' to finish: ")

        # Keep asking until the ID is valid and unique,
        # or the professor enters 'end'.
        while True:

            # Check whether the professor wants to stop.
            if student_id.lower() == "end":
                break

            # Validate that the ID contains exactly two digits.
            if not (len(student_id) == 2 and student_id.isdigit()):

                # Display the required coursework warning.
                print("The input you entered was invalid")

                # Explain the specific problem.
                print("Student ID must contain exactly two digits.")

            else:

                # Extract IDs from previously stored records.
                existing_ids = [
                    student["Student_id"] for student in students
                ]

                # Check whether the ID already exists.
                if student_id in existing_ids:

                    # Display the required coursework warning.
                    print("The input you entered was invalid")

                    # Explain why the ID was rejected.
                    print("This student ID already exists.")

                else:

                    # Accept the valid, unique ID.
                    break

            # Ask again after an invalid or duplicate ID.
            student_id = input("Enter student ID (2 digits), or 'end' to finish: ")

        # Stop the outer for loop if 'end' was entered.
        if student_id.lower() == "end":
            break

        # Display the heading only for an accepted student.
        print(f"\nStudent {student_number + 1}\n")

        # Display the accepted student ID.
        print("Student ID:", student_id)

    # PART 3J - STUDENT NAME

        # Call the get_student_name() function to request
        # and validate the student's name.
        # Store the returned name in the student_name variable.   
        student_name = get_student_name()

        # Display the formatted student name.
        print("Student Name:", student_name)


        # PART 3K - STUDENT DATE OF BIRTH AND AGE

        # Call the date-of-birth validation function.
        # Store the returned date object in student_dob.
        student_dob = get_student_dob()

        # Pass the validated date into the age-calculation function.
        # Store the returned age in student_age.
        student_age = calculate_student_age(student_dob)

        # PART 3E - ASSESSMENT SCORE VALIDATION

        # Request and validate Coursework 1.
        coursework1 = get_valid_score("Enter Coursework 1 score: ")
        print("Coursework 1 score:", coursework1)

        # Request and validate Coursework 2.
        coursework2 = get_valid_score("Enter Coursework 2 score: ")
        print("Coursework 2 score:", coursework2)

        # Request and validate Coursework 3.
        coursework3 = get_valid_score("Enter Coursework 3 score: ")
        print("Coursework 3 score:", coursework3)

        # Request and validate the Final Exam.
        final_exam = get_valid_score("Enter Final Exam score: ")
        print("Final Exam score:", final_exam)

        # PART 3F - CALCULATE THE OVERALL SCORE

        # Pass four assessment scores to the calculation function.
        overall_score = calculate_overall_score(
            coursework1,
            coursework2,
            coursework3,
            final_exam
        )

        # Display the overall score with two decimal places.
        print(f"Final score: {overall_score:.2f}")

        # PART 3C - ROUND THE SCORE AND DETERMINE CATEGORY

        #TOUPLE RETURN

        # Receive both values returned by round_to_category().
        rounded_mark, category = round_to_category(overall_score)

        # Display the rounded mark.
        print("Rounded university mark:", rounded_mark)

        # Display the corresponding category.
        print("Student category:", category)


        # PART 3D - STORE THE STUDENT RECORD

        # Create a dictionary for the current student.
        # Student personal information
        student = {
            "Student_id": student_id,
            "Student_name": student_name,
            "Date_of_birth": student_dob.isoformat(),
            "Age": student_age,
            "Overall_score": round(overall_score, 2),
            "Rounded_mark": rounded_mark,
            "Category": category
        }

        # Add the student dictionary to the list.
        students.append(student)

    # Sort the records by the two-digit student ID.
    students.sort(key=lambda student: student["Student_id"])
  
    # PART 3N - DISPLAY STUDENT RESULTS AS A TABLE

    # Sort the student records by ID in ascending order.
    students.sort(key=lambda student: student["Student_id"])

    # Define the seven required table column headings.
    headers = [
        "UID",
        "Name",
        "D.o.B",
        "Age",
        "Raw Score",
        "Rounded Score",
        "Category"
    ]

    # Create an empty list to hold the table rows.
    rows = []

    # Process each student dictionary in the sorted list.
    for student in students:

        # Extract the student's values in the same order as the headers.
        row = [
            student["Student_id"],
            student["Student_name"],
            student["Date_of_birth"],
            student["Age"],
            student["Overall_score"],
            student["Rounded_mark"],
            student["Category"]
        ]

        # Add the completed row to the list of table rows.
        rows.append(row)

    # Use tabulate to convert the rows into a formatted table.
    student_table = tabulate(rows, headers=headers, tablefmt="grid", disable_numparse=True)

    # Display the completed table.
    print("\nStudent Results:")
    print(student_table)

    # PART 3O - SAVE STUDENT RESULTS TO A TEXT FILE

    # Open or create students.txt in write mode.
    # The with statement closes the file automatically.
    with open("students.txt", "w", encoding="utf-8") as file:

        # Write the formatted student table into the file.
        file.write(student_table)

    # Confirm that the program finished writing the report.
    print("\nStudent results saved to students.txt")


    # Display the total number of stored student records.
    print(f"\nTotal students processed: {len(students)}")


# Start the Student Grading System.
main()
```
