```python

# STUDENT GRADING SYSTEM
# PART 3C - DEFINE THE CATEGORICAL ROUNDING FUNCTION

# Define a function that receives the student's overall score.
def round_to_category(score):

    # Store all permitted university marks.
    allowed_marks = [
        0, 5, 15, 25, 32, 35, 38,
        42, 45, 48, 52, 55, 58,
        62, 65, 68, 72, 75, 78,
        82, 85, 92, 100
    ]

    # Find the permitted mark closest to the student's score.
    # abs() measures the distance between two numbers.
    # min() selects the candidate with the smallest distance.
    # -candidate_mark selects the higher mark when distances are equal.
    nearest_mark = min(
        allowed_marks,
        key=lambda candidate_mark: (
            abs(score - candidate_mark),
            -candidate_mark
        )
    )

    # Return the nearest permitted mark.
    return nearest_mark
```

```python

# PART 3B - DETERMINE THE STUDENT'S GRADING CATEGORY

# Define a function that determines the category from a rounded mark.
def determine_category(rounded_mark):
    
# The rounded_mark variable was calculated in Cell 2.
# Check which category contains that permitted mark.

# Check for Gold Standard.
    if rounded_mark == 100:
        print("Gold Standard")

# Check for Upper First.
    elif rounded_mark == 82 or rounded_mark == 85 or rounded_mark == 92:
        print("Upper First")

# Check for First.
    elif rounded_mark == 72 or rounded_mark == 75 or rounded_mark == 78:
        print("First")

# Check for Upper Second.
    elif rounded_mark == 62 or rounded_mark == 65 or rounded_mark == 68:
        print("Upper Second")

# Check for Lower Second.
    elif rounded_mark == 52 or rounded_mark == 55 or rounded_mark == 58:
        print("Lower Second")

# Check for Third.
    elif rounded_mark == 42 or rounded_mark == 45 or rounded_mark == 48:
        print("Third")

# Check for Condonable Fail.
    elif rounded_mark == 32 or rounded_mark == 35 or rounded_mark == 38:
        print("Condonable Fail")

# Check for Fail.
    elif rounded_mark == 5 or rounded_mark == 15 or rounded_mark == 25:
        print("Fail")

# Check for No Submission.
    elif rounded_mark == 0:
        print("No Submission")

# If no permitted mark matches, display an error message.
    else:
        print("Invalid Mark")
```

```python

# PART 3E - ASSESSMENT SCORE VALIDATION

# Define a reusable function for entering assessment scores.
def get_valid_score(message):

    # Keep asking untill a valid score is returned.
    while True:
        try:
            # Ask for a score and convert it into a decimal number.
            score = float(input(message))

            # Accept and return the score if it is between 0 and 100.
            if 0 <= score <= 100:
                # Return the accepted score.
                return score
            print("The score you entered was invalid. Please enter a number between 0 and 100.")
        except ValueError:
            # Handle text that cannot be converted into a number.
            print("Letter characters are not allowed. Please enter a numeric value between 0 and 100.")
          
  
```

```python
 
# PART 3F - CALCULATE THE OVERALL SCORE

# Define a function that receives four assessment scores.
def calculate_overall_score(coursework1, coursework2, coursework3, final_exam):

    # Multiply each score by its assessment weight.
    overall_score = (
        coursework1 * 0.10 + coursework2 * 0.20
        + coursework3 * 0.30 + final_exam * 0.40
        )

    # Return the calculated score to the calling code.
    return overall_score
```

```python

# PART 3J - STUDENT FULL NAME VALIDATION

# Define a reusable function to collect and validate a full name.
def get_student_name():

    # Keep requesting a name until at least two words are entered.
    while True:

        # Ask for the student's full name.
        # strip() removes spaces at the beginning and end.
        student_name = input("Enter student's full name: ").strip()

        # split() separates the name into individual words.
        name_parts = student_name.split()

        # Accept the name if it contains at least two words.
        if len(name_parts) >= 2:

            # Capitalise the first letter of each word.
            student_name = " ".join(name_parts).title()

            # Return the validated full name to main().
            return student_name

        # Display a warning if fewer than two words were entered.
        print("Please enter the student's full name. ")
```

```python

# PART 3G / 3H / 3I - MAIN STUDENT GRADING PROGRAM

def main():

    # Create an empty list to store all student records.
    students = []

    # Process a maximum of three students.
    for student_number in range(3):

        # Display the current student number.
        print(f"\nStudent {student_number + 1}")

        # Ask for a student ID or the command 'end'.
        student_id = input(
            "Enter student ID (2 digits), or 'end' to finish: "
        )

        # Keep asking until the ID is valid and unique,
        # or the professor enters 'end'.
        while True:

            # Check whether the professor wants to stop.
            if student_id.lower() == "end":
                break

            # Validate that the ID contains exactly two digits.
            if not (len(student_id) == 2 and student_id.isdigit()):

                # Display an invalid-ID warning.
                print(
                    "The student ID you entered was invalid. "
                    "Please enter exactly 2 digits."
                )

            else:

                # Extract IDs from previously stored records.
                existing_ids = [
                    student["Student_id"] for student in students
                ]

                # Check whether the ID already exists.
                if student_id in existing_ids:

                    # Reject duplicate IDs.
                    print("This student ID already exists.")

                else:

                    # Accept the valid, unique ID.
                    break

            # Ask again after an invalid or duplicate ID.
            student_id = input(
                "Enter student ID (2 digits), or 'end' to finish: "
            )

        # Stop the outer for loop if 'end' was entered.
        if student_id.lower() == "end":
            break

        # Display the accepted student ID.
        print("Student ID:", student_id)

    # PART 3J - STUDENT NAME

        # Call the get_student_name() function to request
        # and validate the student's name.
        # Store the returned name in the student_name variable.   
        student_name = get_student_name()
        print("Student Name:", student_name)

        # PART 3E - ASSESSMENT SCORE VALIDATION

        # Request and validate Coursework 1.
        coursework1 = get_valid_score(
            "Enter Coursework 1 score: "
        )
        print("Coursework 1 score:", coursework1)

        # Request and validate Coursework 2.
        coursework2 = get_valid_score(
            "Enter Coursework 2 score: "
        )
        print("Coursework 2 score:", coursework2)

        # Request and validate Coursework 3.
        coursework3 = get_valid_score(
            "Enter Coursework 3 score: "
        )
        print("Coursework 3 score:", coursework3)

        # Request and validate the Final Exam.
        final_exam = get_valid_score(
            "Enter Final Exam score: "
        )
        print("Final Exam score:", final_exam)

        # PART 3F - CALCULATE THE OVERALL SCORE

        # Pass four assessment scores to the calculation function.
        overall_score = calculate_overall_score(
            coursework1,
            coursework2,
            coursework3,
            final_exam
        )

        # Display the overall score with two decimal places.
        print(
            f"The final score for module is: {overall_score:.2f}"
        )

        # PART 3C - ROUND THE SCORE

        # Find the nearest permitted university mark.
        rounded_mark = round_to_category(overall_score)

        # Display the rounded mark.
        print("Rounded university mark:", rounded_mark)

        # PART 3B - DETERMINE THE GRADING CATEGORY

        # Display the grading category.
        determine_category(rounded_mark)

        # PART 3D - STORE THE STUDENT RECORD

        # Create a dictionary for the current student.
        # Student personal information
        student = {
            "Student_id": student_id,
            "Student_name": student_name,
            "Overall_score": round(overall_score, 2),
            "Rounded_mark": rounded_mark
        }

        # Add the student dictionary to the list.
        students.append(student)

    # This section is outside the for loop.
    # It runs after three students or an early exit.

    # Display all collected student records.
    print("\nAll student records:")

    # Sort the records by the two-digit student ID.
    students.sort(key=lambda student: student["Student_id"])

    # Display the sorted list.
    for student in students:
        print(student)

    # Display the total number of stored student records.
    print(f"\nTotal students processed: {len(students)}")


# Start the Student Grading System.
main()
```
